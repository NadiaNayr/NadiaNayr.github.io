```python
import requests
import matplotlib.pyplot as plt
print("hello")
```

```python
base_url = "https://earthquake.usgs.gov/fdsnws/event/1/query"

full_url = base_url + "?format=geojson"

print(full_url)
```

```python
response = requests.get(full_url)

data = response.json() 
type(data)
data['metadata']['generated']

records = data["features"]

mags = []

for r in records:
    mag = r["properties"]["mag"]
    mags.append(mag)

len(mags)

plt.hist(mags)
plt.show()
```
