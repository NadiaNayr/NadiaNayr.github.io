```python
import requests
import matplotlib.pyplot as plt
print("hello")
```

```python
base_url = "https://earthquake.usgs.gov/fdsnws/event/1/query"

full_url = base_url + "?format=csv"

print(full_url)
```

```python
response = requests.get(full_url)
```

```python
response.text.splitlines()[1]
```

```python
headers = response.text.splitlines()[0]
records = response.text.splitlines()[1:]

len(records)

magnitudes = []

for r in records:
    magnitudes.append(float(r.split(",")[4]))

max(magnitudes)
```

## graph thingy

```python
plt.hist(magnitudes)
plt.show()
```
